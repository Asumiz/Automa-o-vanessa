```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager

service = Service(ChromeDriverManager().install())
nav = webdriver.Chrome(service=service)
nav.get("https://web.whatsapp.com/")

#conecta com o whatsapp web e espera o usuário escanear o QR code
```

```python
from selenium.webdriver.common.keys import Keys
import pyperclip
import time

mensagem = """
Teste de envio de mensagem automatizado para o WhatsApp!
"""
lista_contatos = ["Ezequiel", "Meu numero", "Jessica","Morrão Family", "Simone"]

#enviar mensagem para o meu número e depois encaminhas para os outros contatos
# clicar na lupa
nav.find_element('xpath', '//*[@id="side"]/div[1]/div[1]/div/div/div/div/div/div[1]/span').click()
time.sleep(1)

#escrever o nome na barra de pesquisa
nav.find_element('xpath', '//*[@id="_r_a_"]').send_keys("Ezequiel")
nav.find_element('xpath', '//*[@id="_r_a_"]').send_keys(Keys.ENTER)

#escrever mensagem 
pyperclip.copy(mensagem)
nav.find_element('xpath', '//*[@id="main"]/footer/div[1]/div/span/div/div/div/div[3]/div[1]/p').send_keys(Keys.CONTROL + "v")
nav.find_element('xpath', '//*[@id="main"]/footer/div[1]/div/span/div/div/div/div[3]/div[1]/p').send_keys(Keys.ENTER)
time.sleep(2)
```
